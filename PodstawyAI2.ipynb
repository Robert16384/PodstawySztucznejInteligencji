```python
import math
import random
import time
import matplotlib.pyplot as plt

random.seed(0)

LINIE = [(0, 1, 2), (3, 4, 5), (6, 7, 8),   # wiersze
         (0, 3, 6), (1, 4, 7), (2, 5, 8),   # kolumny
         (0, 4, 8), (2, 4, 6)]              # przekątne

class KolkoKrzyzyk:

    start = ('.',) * 9          # 9 wolnych pól

    def gracz(self, s):
        if s.count('X') == s.count('O'):
            return "MAX"
        else:
            return "MIN"

    def akcje(self, s):
        wolne = []
        for i in range(9):
            if s[i] == '.':
                wolne.append(i)
        return wolne

    def wynik(self, s, a):
        if self.gracz(s) == "MAX":
            znak = 'X'
        else:
            znak = 'O'
        # nowa krotka: pola przed a, nowy znak, pola po a
        return s[:a] + (znak,) + s[a + 1:]

    def zwyciezca(self, s):
        for a, b, c in LINIE:
            if s[a] != '.' and s[a] == s[b] == s[c]:
                return s[a]          # 'X' albo 'O'
        return None                  # nikt nie wygrał

    def koniec(self, s):
        return self.zwyciezca(s) is not None or '.' not in s

    def wyplata(self, s):
        z = self.zwyciezca(s)
        if z == 'X':
            return 1
        if z == 'O':
            return -1
        return 0


kk = KolkoKrzyzyk()


def gracz_losowy(gra, s):
    return random.choice(gra.akcje(s))


def rozegraj(gra, gracz_max, gracz_min, pokaz=True):
    s = gra.start
    while not gra.koniec(s):
        if gra.gracz(s) == "MAX":
            ruch = gracz_max(gra, s)
        else:
            ruch = gracz_min(gra, s)
        s = gra.wynik(s, ruch)
    return s


rozegraj(kk, gracz_losowy, gracz_losowy)


wygrane_x = 0
wygrane_o = 0
remisy = 0

# TODO: 1000 razy rozegraj partię i zlicz wyniki

for i in range(1000):
    s = rozegraj(kk, gracz_losowy, gracz_losowy)

    wynik = kk.wyplata(s)

    if wynik == 1:
        wygrane_x += 1
    elif wynik == -1:
        wygrane_o += 1
    else:
        remisy += 1
    print("X:", wygrane_x, " O:", wygrane_o, " remisy:", remisy)
```
